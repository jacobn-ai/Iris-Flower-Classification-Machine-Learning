Prepare X and Y

```python
X = df[
    [
        "sepal_length",
        "sepal_width",
        "petal_length",
        "petal_width"
    ]
]

y = df["species"]
```

```python
print("X shape:", X.shape)
print("y shape:", y.shape)
```

```python
from sklearn.preprocessing import LabelEncoder

label_encoder = LabelEncoder()

y_encoded = label_encoder.fit_transform(y)

print(y_encoded[:10])
```

```python
for number, name in enumerate(label_encoder.classes_):
    print(number, "=", name)
```

Train/Test Split

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y_encoded,
    test_size=0.20,
    random_state=42,
    stratify=y_encoded
)
```

```python
print("Training samples:", X_train.shape[0])
print("Testing samples:", X_test.shape[0])
```

Feature Scaling

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)

X_test_scaled = scaler.transform(X_test)
```

```python
y_pred = logistic_model.predict(
    X_test_scaled
)
```

Classification report

```python
from sklearn.tree import plot_tree

plt.figure(figsize=(20,10))

plot_tree(
    decision_tree,
    feature_names=X.columns,
    class_names=label_encoder.classes_,
    filled=True,
    rounded=True
)

plt.title("Decision Tree for Iris Classification")
plt.show()
```

```python
depths = range(1, 11)

train_accuracies = []
test_accuracies = []

for depth in depths:

    model = DecisionTreeClassifier(
        max_depth=depth,
        random_state=42
    )

    model.fit(X_train, y_train)

    train_pred = model.predict(X_train)
    test_pred = model.predict(X_test)

    train_accuracies.append(
        accuracy_score(y_train, train_pred)
    )

    test_accuracies.append(
        accuracy_score(y_test, test_pred)
    )
```

```python
plt.figure(figsize=(8,5))

plt.plot(
    depths,
    train_accuracies,
    marker="o",
    label="Training Accuracy"
)

plt.plot(
    depths,
    test_accuracies,
    marker="o",
    label="Testing Accuracy"
)

plt.xlabel("Maximum Tree Depth")
plt.ylabel("Accuracy")
plt.title("Decision Tree: Depth vs Accuracy")

plt.xticks(list(depths))
plt.legend()
plt.grid(True)

plt.show()
```

```python
for depth, train_acc, test_acc in zip(
    depths,
    train_accuracies,
    test_accuracies
):
    print(
        f"Depth={depth}: "
        f"Train={train_acc:.4f}, "
        f"Test={test_acc:.4f}"
    )
```

```python
decision_tree_tuned = DecisionTreeClassifier(
    max_depth=3,
    random_state=42
)

decision_tree_tuned.fit(X_train, y_train)

y_pred_tree_tuned = decision_tree_tuned.predict(X_test)

tree_tuned_accuracy = accuracy_score(
    y_test,
    y_pred_tree_tuned
)

print("Tuned Decision Tree Accuracy:", tree_tuned_accuracy)
```

K-Means clustering

```python
from sklearn.cluster import KMeans

kmeans = KMeans(
    n_clusters=3,
    random_state=42,
    n_init=10
)

clusters = kmeans.fit_predict(X_train_scaled)

print("Cluster assignments:")
print(clusters)

print("\nCluster Centers:")
print(kmeans.cluster_centers_)
```

```python
import matplotlib.pyplot as plt

inertias = []

for k in range(1, 11):

    km = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )

    km.fit(X_train_scaled)

    inertias.append(km.inertia_)

plt.figure(figsize=(8, 5))

plt.plot(
    range(1, 11),
    inertias,
    marker="o"
)

plt.xlabel("Number of Clusters (k)")
plt.ylabel("Inertia")
plt.title("Elbow Method for K-Means")

plt.xticks(range(1, 11))
plt.grid()

plt.show()
```

```python
from sklearn.metrics import silhouette_score

silhouette_scores = []

for k in range(2, 11):

    km = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )

    cluster_labels = km.fit_predict(X_train_scaled)

    score = silhouette_score(
        X_train_scaled,
        cluster_labels
    )

    silhouette_scores.append(score)

    print(
        f"k = {k}, Silhouette Score = {score:.4f}"
    )
```

```python
plt.figure(figsize=(8, 5))

plt.plot(
    range(2, 11),
    silhouette_scores,
    marker="o"
)

plt.xlabel("Number of Clusters (k)")
plt.ylabel("Silhouette Score")
plt.title("Silhouette Score for K-Means")

plt.xticks(range(2, 11))
plt.grid()

plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.plot(
    range(2, 11),
    silhouette_scores,
    marker="o"
)

plt.xlabel("Number of Clusters (k)")
plt.ylabel("Silhouette Score")
plt.title("Silhouette Score for K-Means")

plt.xticks(range(2, 11))
plt.grid()

plt.show()
```

Pipeline

```python
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.svm import SVC

svm_pipeline = Pipeline([
    ("scaler", StandardScaler()),
    ("model", SVC(kernel="linear", random_state=42))
])

svm_pipeline.fit(X_train, y_train)

y_pred_pipeline = svm_pipeline.predict(X_test)
```

```python
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

pipeline_accuracy = accuracy_score(
    y_test,
    y_pred_pipeline
)

print(
    "Pipeline Accuracy: {:.2f}%".format(
        pipeline_accuracy * 100
    )
)

print("\nClassification Report:")
print(
    classification_report(
        y_test,
        y_pred_pipeline
    )
)

print("\nConfusion Matrix:")
print(
    confusion_matrix(
        y_test,
        y_pred_pipeline
    )
)
```

CV Proper way

```python
from sklearn.model_selection import cross_val_score

cv_scores = cross_val_score(
    svm_pipeline,
    X_train,
    y_train,
    cv=5,
    scoring="accuracy"
)

print("Cross-validation scores:")
print(cv_scores)

print("\nMean CV Accuracy: {:.2f}%".format(
    cv_scores.mean() * 100
))

print("CV Standard Deviation: {:.2f}%".format(
    cv_scores.std() * 100
))
```
